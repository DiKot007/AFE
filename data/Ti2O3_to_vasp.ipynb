```python
from pymatgen.core import Structure

# Загружаем структуру из скачанного ранее CIF-файла mp-458
structure = Structure.from_file("Vampire_data/Ti2O3.cif")

# Экспортируем в формат POSCAR для VASP
structure.to(filename="VASP_data/POSCAR", fmt="poscar")
print("Файл POSCAR успешно создан!")
```
